- 它可以在 CPU、NVIDIA GPU（`"cuda"`）、Apple Silicon GPU（`"mps"`）和 Intel GPU（`"xpu"`）上正常运行；本章后面会进一步讨论性能取舍。

```python
def get_device(enable_tensor_cores=True):
    if torch.cuda.is_available():
        device = torch.device("cuda")
        print("Using NVIDIA CUDA GPU")
        
        if enable_tensor_cores:
            major, minor = map(int, torch.__version__.split(".")[:2])
            # PyTorch 2.9 and 2.10 still read the legacy TF32 setting in torch.compile.
            # See https://github.com/pytorch/pytorch/issues/166387
            # and https://github.com/rasbt/reasoning-from-scratch/issues/256
            if (major, minor) >= (2, 11):
                torch.backends.cuda.matmul.fp32_precision = "tf32"
                torch.backends.cudnn.conv.fp32_precision = "tf32"
            else:
                torch.backends.cuda.matmul.allow_tf32 = True
                torch.backends.cudnn.allow_tf32 = True

    elif torch.backends.mps.is_available():
        device = torch.device("mps")
        print("Using Apple Silicon GPU (MPS)")

    elif torch.xpu.is_available():
        device = torch.device("xpu")
        print("Using Intel GPU")

    else:
        device = torch.device("cpu")
        print("Using CPU")

    return device

device = get_device()
```

## 2.7 编写最小文本生成函数

<img src="https://sebastianraschka.com/images/reasoning-from-scratch-images/ch02/CH02_F13_raschka.webp" width="500px">

<img src="https://sebastianraschka.com/images/reasoning-from-scratch-images/ch02/CH02_F14_raschka.webp?2" width="500px">

- `generate_text_basic_stream` 函数实现了这个顺序文本生成过程：

```python
@torch.inference_mode()
def generate_text_basic_stream(
    model,
    token_ids,
    max_new_tokens, 
    eos_token_id=None
):
    model.eval()

    for _ in range(max_new_tokens):
        out = model(token_ids)[:, -1]
        next_token = torch.argmax(out, dim=-1, keepdim=True)

        # Stop if we encounter an end-of-sequence token
        if (eos_token_id is not None
                and torch.all(next_token == eos_token_id)):
            break

        yield next_token  # Yield each token as it's generated
        
        token_ids = torch.cat([token_ids, next_token], dim=1)
```

- 让我们使用它，根据简单的提示词 `"Explain large language models in a single sentence."` 生成一段 100 token 的回答，看看它如何工作（推理相关内容会在后续章节介绍）。
- 下面的代码运行较慢，根据计算机性能可能需要 1–3 分钟才能完成（后续小节会加速它）。

```python
prompt = "Explain large language models in a single sentence."
input_token_ids_tensor = torch.tensor(
    tokenizer.encode(prompt),
    device=device
    ).unsqueeze(0)
max_new_tokens = 100


for token in generate_text_basic_stream(
    model=model,
    token_ids=input_token_ids_tensor,
    max_new_tokens=max_new_tokens,
):
    token_id = token.squeeze(0).tolist()
    print(
        tokenizer.decode(token_id),
        end="",
        flush=True  # Deactivates buffering so tokens are printed live
    )
```

- 注意，LLM 能够很好地遵循指令，但在 `<|endoftext|>` 之后，回答会变得无意义或偏离主题；这个 token 在训练期间用于分隔不同文档。
- 使用 LLM 时，我们希望它遇到这个 token 后停止生成。

```python
print(tokenizer.encode("<|endoftext|>"))
```

- 为了使用方便，这个 token ID 被存储为 tokenizer 的一个属性（eos，即 end of sequence，序列结束）：

```python
print(tokenizer.eos_token_id)
```

- 我们可以利用它告诉 LLM（更准确地说，是告诉 `generate_text_basic_stream` 函数）何时停止生成文本。

```python
for token in generate_text_basic_stream(
    model=model,
    token_ids=input_token_ids_tensor,
    max_new_tokens=max_new_tokens,
    eos_token_id=tokenizer.eos_token_id  # Use EOS token
):
    token_id = token.squeeze(0).tolist()
    print(
        tokenizer.decode(token_id),
        end="",
        flush=True
    )
```

- 上面的回答是在 CPU 上运行得到的；根据设备不同，生成文本可能会略有差异。

- 在结束本节并了解如何加速代码之前，我们先实现一个简单的基准测试函数，用来记录计算性能。

```python
import warnings

def generate_stats(output_token_ids, tokenizer, start_time,
                   end_time):
    # tokenizer is currently unused but retained for backward compatibility
    total_time = end_time - start_time
    print(f"\n\nTime: {total_time:.2f} sec")
    print(f"{int(output_token_ids.numel() / total_time)} tokens/sec")

    for name, backend in (("CUDA", getattr(torch, "cuda", None)),
                          ("XPU", getattr(torch, "xpu", None))):
        if backend is not None and backend.is_available():

            # Check whether we are actually using this backend
            device_type = output_token_ids.device.type
            if device_type != name.lower():
                warnings.warn(
                    f"{name} is available but tensors are on "
                    f"{device_type}. Memory stats may be 0."
                )
    
            # Synchronize if supported (important for async backends)
            if hasattr(backend, "synchronize"):
                backend.synchronize()
            
            max_mem_bytes = backend.max_memory_allocated()
            max_mem_gb = max_mem_bytes / (1024 ** 3)
            print(f"Max {name} memory allocated: {max_mem_gb:.2f} GB")
            backend.reset_peak_memory_stats()
```

```python
import time

start_time = time.time()
generated_ids = []

for token in generate_text_basic_stream(
    model=model,
    token_ids=input_token_ids_tensor,
    max_new_tokens=max_new_tokens,
    eos_token_id=tokenizer.eos_token_id
):
    token_id = token.squeeze(0).tolist()
    print(
        tokenizer.decode(token_id),
        end="",
        flush=True
    )

    next_token_id = token.squeeze(0)
    generated_ids.append(next_token_id)  # Collect generated tokens

end_time = time.time()

output_token_ids_tensor = torch.cat(generated_ids, dim=0)
generate_stats(output_token_ids_tensor, tokenizer, start_time, end_time)
```

## 2.8 通过 KV 缓存加速推理

<img src="https://sebastianraschka.com/images/reasoning-from-scratch-images/ch02/CH02_F15_raschka.webp?2" width="500px">

- 本书的代码强调可读性；如果专门讨论优化，完全可以另写一本书。
- 这里介绍一种叫作“KV 缓存”的工程技巧（KV 指 LLM 注意力机制中的 key 和 value）。
- 如果你不熟悉这些术语，不必担心；只需知道，我们可以把每次迭代都会重复使用的中间值存储（缓存）起来。

<img src="https://sebastianraschka.com/images/reasoning-from-scratch-images/ch02/CH02_F16_raschka.webp" width="500px">

- 关于 KV 缓存机制的更多细节，请参阅我的文章[《从零理解并编写 LLM 中的 KV Cache》](https://magazine.sebastianraschka.com/p/coding-the-kv-cache-in-llms)。
- 下面是使用 KV 缓存改写后的 `generate_text_basic_stream` 函数。

```python
from reasoning_from_scratch.qwen3 import KVCache

@torch.inference_mode()
def generate_text_basic_stream_cache(
    model,
    token_ids,
    max_new_tokens,
    eos_token_id=None
):
    model.eval()
    cache = KVCache(n_layers=model.cfg["n_layers"])  # New
    model.reset_kv_cache()                           # New

    out = model(token_ids, cache=cache)[:, -1]
    for step in range(max_new_tokens):
        next_token = torch.argmax(out, dim=-1, keepdim=True)

        if (eos_token_id is not None
                and torch.all(next_token == eos_token_id)):
            break

        yield next_token
        if step + 1 < max_new_tokens:  # Skips the forward pass after the last token is generated
            out = model(next_token, cache=cache)[:, -1]
```

- 使用方式与之前类似：

```python
start_time = time.time()
generated_ids = []

for token in generate_text_basic_stream_cache(
    model=model,
    token_ids=input_token_ids_tensor,
    max_new_tokens=max_new_tokens,
    eos_token_id=tokenizer.eos_token_id
):
    token_id = token.squeeze(0).tolist()
    print(
        tokenizer.decode(token_id),
        end="",
        flush=True
    )

    next_token_id = token.squeeze(0)
    generated_ids.append(next_token_id)  # Collect generated tokens

end_time = time.time()

output_token_ids_tensor = torch.cat(generated_ids, dim=0)
generate_stats(output_token_ids_tensor, tokenizer, start_time, end_time)
```

- 可以看到，在 Mac Mini M4 CPU 上，KV 缓存让生成速度显著提升。

**Windows 说明 1**

- 在 Windows 上进行编译可能比较棘手。
- `torch.compile()` 使用 Inductor，它会即时编译 kernel，因此需要可用的 C/C++ 工具链。
- 对 CUDA 而言，Inductor 还依赖 Triton；Windows 上可以使用社区维护的 `triton-windows` 包。
  - 如果看到 `cl not found`，请[安装带有“C++ 工作负载”的 Visual Studio Build Tools](https://learn.microsoft.com/en-us/cpp/build/vscpp-step-0-installation?view=msvc-170)，并从“x64 Native Tools”命令提示符运行 Python。
  - 如果使用 CUDA 时看到 `triton not found`，请安装 `triton-windows`（例如 `uv pip install "triton-windows<3.4"`）。
- 对 CPU 而言，有读者进一步推荐这份 [Windows PyTorch Inductor 指南](https://docs.pytorch.org/tutorials/unstable/inductor_windows.html)。
  - 安装 Visual Studio 2022 时，务必安装英文语言包，以避免 UTF-8 错误。
  - 另请注意，代码需要通过“Visual Studio 2022 Developer Command Prompt”运行，不能从 Notebook 运行。
- 如果配置过程过于复杂，可以跳过编译；**编译是可选的，所有代码示例不编译也能正常运行。**

**Windows 说明 2**

- 有读者报告，在 Windows 默认设置下运行 `torch.compile` 没有加速效果；但使用 `"max-autotune"` 模式后可以获得约 2 倍加速：`torch.compile(model, mode="max-autotune")`
